```python
import pandas as pd
stock = pd.read_csv("~/Downloads/stockdata.csv")
print(stock.head())
print(stock.shape)
print(stock.info())
```

```python
#Visualizziamo l'andamento delle azioni della Apple ("AAPL")
stock['Date'] = pd.to_datetime(stock['Date']) #mi trasformo formato Date da str a datetime
sns.lineplot(data= stock, x = 'Date', y = 'AAPL')
```

```python
# • Visualizziamo l'andamento di tutte le azioni del dataset • Visualizziamo l'andamento di Microsoft ("MSFT") e IBM ("IBM") negli anni 2012 e 2013 • Con un istogramma, troviamo il range di valori che le azioni di Microsoft hanno assunto più spesso
stock.plot(x='Date', y=['MSFT', 'IBM', 'SBUX', 'AAPL', 'GSPC'])
```

```python
#Visualizziamo l'andamento di Microsoft ("MSFT") e IBM ("IBM") negli anni 2012 e 2013
inizio_2012 = '2012-01-01'
fine_2013 = '2013-12-31'
stock_2012_2013 = stock.loc[(stock['Date'] >= inizio_2012) & (stock['Date'] <= fine_2013)]

stock_2012_2013.plot( x = 'Date', y = ['MSFT', 'IBM'])
```

```python
#Con un istogramma, troviamo il range di valori che le azioni di Microsoft hanno assunto più spesso
stock['MSFT'].plot(kind = 'hist', bins = 20)
```

```python
import pandas as pd
metal = pd.read_csv("~/Desktop/Python - Epicode/datasets/altri_dataset/metal_bands_2017.csv", encoding = "latin1", index_col = 0)
metal
```

```python
#Istogramma del numero dei fan delle varie band; cosa si vede? Cose ne deduciamo? 
metal.plot(x = 'band_name', y = 'fans', kind = 'hist', bins = 50)

print("Ci accorgiamo che pochissime band hanno tantissimi fan")
```

```python
#Delle band che suonano sia Progressive che Thrash, realizziamo un barplot che mostri quanti fan ci sono in totale rispetto al paese di origine
filtro_prog = metal['style'].str.contains('progressive', case = False)
filtro_thrash = metal['style'].str.contains('thrash', case = False)

filtro = filtro_prog & filtro_thrash

prog_thrash = metal.loc[filtro]

prog_thrash.groupby('origin')['fans'].size().plot(kind = 'barh')
```

```python
#Visualizziamo un istogramma delle date di formazione delle band (potremo dover convertire in qualche modo questo dato); si nota qualche trend?

metal.dtypes

filtro = metal['formed'] != "-"
metal_f = metal.loc[filtro]

metal_f['formed']= metal_f['formed'].astype(int)

metal_f['formed'].plot(kind = 'hist', bins = 50)

print("Si nota un trend in crescita con un picco tra il 2000 e il 2010 per poi notare una discesa progressiva")
```

```python
import pandas as panda
import seaborn as sns
party = pd.read_csv("datasets/altri_dataset/party_in_nyc.csv")
party
```

```python
#Realizzo un grafico che mostra quante chiamate ci sono state per ogni città per tutti i tipi di locale
sns.countplot(data = party, y = 'Location Type')
```

```python
#Qual è il rettangolo di latitudine e longitudine considerato dal dataset?
lat_min = party["Latitude"].min()
lat_max = party["Latitude"].max()
lon_min = party["Longitude"].min()
lon_max = party["Longitude"].max()

print("Latitudine: da", lat_min, "a", lat_max)
print("Longitudine: da", lon_min, "a", lon_max)
```

```python
 #Quante segnalazioni sono state fatte per case private? 
house_calls = party[party["Location Type"] == "Residential Building/House"]

print("Segnalazioni per case private:", len(house_calls))
```

```python
#Visualizzo il conteggio delle segnalazioni avvenute in ogni quartiere
sns.countplot(data=party, y='Borough', hue='Location Type')
```
