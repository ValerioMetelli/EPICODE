```python
import pandas as pd

# Carico i 2 dataset
festival = pd.read_excel('~/EPICODE/Progetto Finale Looker/drive-download-20250804T194039Z-1-001/dati-festival-sanremo-1951-2023.xlsx')

classifica = pd.read_excel('~/EPICODE/Progetto Finale Looker/drive-download-20250804T194039Z-1-001/Ufficiale/dati-classifica-sanremo-1951-2023.xlsx')

display(festival.head(), classifica.head())
```

```python
print(f'FESTIVAL: {festival.shape}')
print(f'CLASSIFICA: {classifica.shape}')

print('\nCOLONNE FESTIVAL:\n', list(festival.columns))
print('\nCOLONNE CLASSIFICA:\n', list(classifica.columns))

print(F'\nFESTIVAL INFO: \n{festival.info}\nCLASSIFICA INFO: \n{classifica.info}')
```

```python
#Modifico in minuscolo i nomi di tutte le colonne per comodità
festival.columns = festival.columns.str.strip().str.lower()
classifica.columns = classifica.columns.str.strip().str.lower()
```

***

### ANALISI E MODIFICHE TABELLA 'CLASSIFICA'

```python
#controllo se ci sono duplicati
print('Duplicati in classifica:', classifica.duplicated().sum())

#controllo se ci sono nulli
print('\n','Classifica: \n', classifica.isnull().sum())
```

```python
#Controllo i valori nulli all'interno della colonna "Posizione"
classifica.loc[classifica["posizione"].isna()]
```

```python
#SOSTITUISCO TUTTI I VALORI NULL 

#Trovo i NaN nella colonna Posizione
mask = classifica["posizione"].isna()

#Conto quanti NaN ci sono
n_nan = mask.sum()

#Creo la sequenza che parte da 11 e lunga quanto i NaN
nuovi_valori = range(11, 11 + n_nan)

#Sostituisco i NaN con i nuovi valori
classifica.loc[mask, "posizione"] = nuovi_valori
```

```python
#Elimino la colonna Unnamed: 0
classifica.drop(columns=['unnamed: 0'], inplace=True)
```

```python
#Cambio il formato della colonna anno da int a datetime
classifica['anno'] = pd.to_datetime(classifica['anno'].astype(str), format='%Y')
festival['anno'] = pd.to_datetime(festival['anno'].astype(str), format='%Y')
```

```python
#Elimino i record dove ci sono i valori F/NF in quanto non sono classificabili e non utili ai fini del report
classifica.drop(classifica[classifica['posizione'].isin(['F', 'NF'])].index, inplace=True)

#Tengo solo le cifre dalla colonna 'posizione'
solo_cifre = (classifica['posizione']
       .astype('string')
       .str.replace(r'[^0-9]', '', regex=True)   # elimina tutto ciò che non è una cifra
       .str.strip())

#Converto a int
classifica["posizione"] = pd.to_numeric(solo_cifre, errors="coerce").astype("Int64")

print(classifica['posizione'].dtype)  
classifica.head(5)
```

```python
#Sostituisco tutti i connettori nella colonna 'interprete' con una ',' per semplificare l'estrazione dei vari nomi

#Creo un dizionario per sostituiri i vari connettori con una ',' 
sostituzioni = {
    r'(?i)\be\b' : ',',
    r'(?i)\bed\b' : ',',
    r'(?i)\bcon\b' : ',',
    r'(?i)\bfeat\.?\b' : ',',
    r'(?i)\bfeaturing\b' : ',',
    r'&' : ',',
    r'/' : ',',
    r'-' : ','
}

#Sostituisco i connettori
classifica['interprete'] = classifica['interprete'].replace(sostituzioni, regex=True).str.strip()

#Creo una lista che mi servirà dopo
classifica['interprete_lista'] = classifica['interprete'].str.split(',')

#Creo x colonne per x interpreti
classifica[[
    'interprete_1', 
    'interprete_2', 
    'interprete_3', 
    'interprete_4',
    'interprete_5'
]] = pd.DataFrame(classifica['interprete_lista'].to_list(), index=classifica.index)
```

```python
#replico lo stesso per "autore"

classifica['autori'] = classifica['autori'].replace(sostituzioni, regex=True).str.strip()

classifica['autori_lista'] = classifica['autori'].str.split(',')

classifica[[
    'autori_1', 
    'autori_2', 
    'autori_3', 
    'autori_4',
    'autori_5', 
    'autori_6', 
    'autori_7', 
    'autori_8', 
    'autori_9'
]] = pd.DataFrame(classifica['autori_lista'].to_list(), index=classifica.index)
```

```python
#scarico i file lavorati
classifica.to_csv("CLASSIFICA.csv", index=False)
festival.to_csv("PRESENTATORI.csv", index=False)

#file solo autori
autori_exploded = classifica.explode("autori_lista").copy()
autori_exploded.rename(columns={"autori_lista": "autore"}, inplace=True)
#seleziono solo le colonne che mi serviranno
autori_light = autori_exploded[["anno", "canzone", "posizione", "autore"]]
autori_light.to_csv("AUTORI.csv", index=False)

#file solo interpreti
interpreti_exploded = classifica.explode("interprete_lista").copy()
interpreti_exploded.rename(columns={"interprete_lista": "interprete"}, inplace=True)
#seleziono solo le colonne che mi serviranno
interpreti_light = interpreti_exploded[["anno", "canzone", "posizione", "interprete"]]
interpreti_light.to_csv("INTERPRETI.csv", index=False)
```
